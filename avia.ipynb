```python
for col in df_flights.columns:
 # Рассчитываем долю пропущенных значений
  pct_missing = np.mean(df_flights[col].isnull())

 # Выводим название признака и процент пропусков
  print('{} - {}%'.format(col, round(pct_missing*100)))
```

```python
# Формируем новое поле "Статус рейса"
# Если фактическое время прилета отсутствует, считаем рейс отмененным, при наличии - выполненным

df_flights['Статус рейса'] = df_flights.apply(lambda row: 'Отменен' if
pd.isna(row['Дата и время прилета факт']) else 'Выполнен', axis=1)
```

```python
df_flights.head()
```

```python
# Считаем количество выполненых рейсов

completed_flights = (df_flights['Статус рейса'] == 'Выполнен').sum()
completed_flights
```

```python
# Считаем количество отменных рейсов

canceled_flights = (df_flights['Статус рейса'] == 'Отменен').sum()
canceled_flights
```

```python
# Формируем бинарное поле "Регулярность"
# Значение 1 получают рейсы, выполненные без задержки, значение 0 - с задержкой или отмененные

df_flights['Регулярность'] = df_flights.apply(lambda row: 1 if row ['Статус рейса'] == 'Выполнен' and row['Задержка по вылету в минутах']<=0 else 0, axis=1)
```

```python
df_flights.head()
```

```python
# Вычисляем количество регулярных рейсов

regular_flights = (df_flights['Регулярность'] == 1).sum()
regular_flights
```

```python
# Также количество нерегулярных рейсов

irregular_flights = (df_flights['Регулярность'] == 0).sum()
irregular_flights
```

```python
# Выводим описательную статистику

df_flights.describe()
```

```python
# Заполняем пропуски в данных
df_flights['Задержка по вылету в минутах']=df_flights['Задержка по вылету в минутах'].fillna(df_flights['Задержка по вылету в минутах'].mean())
```

```python
df_flights['Задержка по прибытию в минутах']=df_flights['Задержка по прибытию в минутах'].fillna(df_flights['Задержка по прибытию в минутах'].mean())
```

```python
df_flights['Время полета факт']=df_flights['Время полета факт'].fillna(df_flights['Время полета факт'].mean())
```

```python
df_flights['Расстояние']=df_flights['Расстояние'].fillna(df_flights['Расстояние'].mean())
df_flights['Время полета план']=df_flights['Время полета план'].fillna(df_flights['Время полета план'].mean())
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
# Выбираем числовые признаки, которые будут использоваться при кластеризации

X = df_flights[['Задержка по вылету в минутах', 'Задержка по прибытию в минутах', 'Расстояние', 'Время полета план','Время полета факт']]
```

```python
# Масштабируем признаки, чтобы все переменные имели одинаковый масштаб

scaled_df = StandardScaler().fit_transform(X)
```

```python
# Задаем параметры алгоритма K-Means
kmeans_kwargs = {
"init": "random",
"n_init": 10,
"random_state": 1,}

# Рассчитываем значение SSE
sse = []
for k in range(1, 11):
  kmeans = KMeans(n_clusters=k, **kmeans_kwargs)
  kmeans.fit(scaled_df)
  sse.append(kmeans.inertia_)
```

```python
# Визуализируем полученные данные
plt.plot(range(1, 11), sse)
plt.xticks(range(1, 11))
plt.xlabel("Number of Clusters")
plt.ylabel("SSE")
plt.show()
```

```python
# Создаем модель K-Means с количеством кластеров

kmeans = KMeans(init="random", n_clusters=3, n_init=10, random_state=1)
kmeans.fit(scaled_df)
```

```python
# Добавляем информацию о принадлежности каждого объекта к найденному кластеру

X['clustering']=kmeans.labels_
```

```python
X.head()
```

```python
# Визуализируем результаты кластеризации
# Каждая точка соответствует отдельному рейсу, а цвет принадлежность к определенному кластеру

plt.scatter(X['Задержка по вылету в минутах'], X['Расстояние'],
c=X['clustering'], cmap='viridis')
plt.xlabel('Задержка по вылету в минутах')
plt.ylabel('Расстояние')
plt.title('Clustering of Flight Data with Regularity')
plt.show()
```

```python
# Преобразуем столбец с датой вылета в формат datetime
df['Дата вылета план']=pd.to_datetime(df['Дата вылета план'], dayfirst=True, errors='coerce')
```

```python
# Разделяем данные на обучающую и тестовую выборки по временному интервалу

filtered_df = df[(df['Дата вылета план'] >= '2012-12-01') & (df['Дата вылета план'] < '2013-12-01')]
filtered_df_test=df[(df['Дата вылета план'] >= '2013-12-01') & (df['Дата вылета план'] < '2013-12-31')]
```

```python
# Формируем матрицу признаков обучающей модели

x_train=filtered_df[['Задержка по вылету в минутах', 'Задержка по прибытию в минутах', 'Расстояние', 'Время полета план','Время полета факт', 'Количество кресел','Статус рейса']]

# Формируем целевую переменную
y_train=filtered_df[['Регулярность']]
```

```python
# Формируем матрицу признаков и переменную для тестовой выборки
x_test= filtered_df_test[['Задержка по вылету в минутах', 'Задержка по прибытию в минутах', 'Расстояние', 'Время полета план','Время полета факт', 'Количество кресел','Статус рейса']]
y_test= filtered_df_test[['Регулярность']]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error, r2_score
```

```python
# Создаем модель линейной регрессии

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Получаем прогноз модели для тестовой выборки
y_pred = model.predict(x_test)
```

```python
# Оцениваем модельmse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f'Mean Squared Error: {mse}')
print(f'R^2 Score: {r2}')
```

```python
# Выводим коэффициенты модели
print("Коэффициенты модели:", model.coef_)
```

```python
import matplotlib.pyplot as plt
```

```python
# Визуализация предсказаний
plt.figure(figsize=(10, 6))
plt.scatter(range(len(y_test)), y_test, color='blue', label='Истинные значения')
plt.scatter(range(len(y_pred)), y_pred, color='red', label='Предсказанные значения')
plt.xlabel('Наблюдения')
plt.ylabel('Значения')
plt.title('Истинные и предсказанные значения')
plt.legend()
plt.show()
```

```python
# Формируем обучающую выборку и переменную с использованием сокращенного набора признаков
x_train=filtered_df[['Задержка по вылету в минутах', 'Время полета план', 'Время полета факт', 'Количество кресел','Статус рейса']]
y_train=filtered_df[['Регулярность']]
```

```python
# Формируем тестовую выборку и переменную для оценки альтернативной модели
x_test= filtered_df_test[['Задержка по вылету в минутах', 'Время полета план', 'Время полета факт', 'Количество кресел','Статус рейса']]
y_test= filtered_df_test[['Регулярность']]
```

```python
# Создаем и обучаем модель на сокращенном наборе признаков
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Получаем прогноз для тестовой выборки
y_pred = model.predict(x_test)
```

```python
# Оцениваем модель
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f'Mean Squared Error: {mse}')
print(f'R^2 Score: {r2}')
```

```python
# Выводим коэффициенты модели
print("Коэффициенты модели:", model.coef_)
```

```python
# Визуализация предсказаний
plt.figure(figsize=(10, 6))
plt.scatter(range(len(y_test)), y_test, color='blue', label='Истинные значения')
plt.scatter(range(len(y_pred)), y_pred, color='red', label='Предсказанные значения')
plt.xlabel('Наблюдения')
plt.ylabel('Значения')
plt.title('Истинные и предсказанные значения')
plt.legend()
plt.show()
```

Сравнение результатов показало, что исключение части признаков практически не повлияло на качество прогнозирования: значения MSE и коэффициента детерминации R^2 изменились незначительно. Полученные значения коэффициента детерминации свидетельствуют о том, что линейная модель объясняет лишь часть изменчивости данных.
